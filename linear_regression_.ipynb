```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error,root_mean_squared_error,r2_score
```

```python
df = pd.read_csv("/content/car_features (1).csv")
print(df.head(10))
```

```python
# data splitting
x = df.drop(["selling_price"], axis = 1)
y = df["selling_price"]
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size= 0.2, random_state= 42)
```

```python
#scaling
scaler = StandardScaler()
x_train_scaler = scaler.fit_transform(x_train)
x_test_scaler = scaler.transform(x_test)
print(x_train_scaler)
print(x_test_scaler)
```

```python
model = LinearRegression()
model_fit = model.fit(x_train,y_train)
print(model_fit)
```

```python
#prediction
y_train_predict = model.predict(x_train)
y_test_predict = model.predict(x_test)
print(y_train_predict)
print(y_test_predict)
```

```python
mae = mean_absolute_error(y_test, y_test_predict)
print(mae)
```

```python
mse_score = mean_squared_error(y_test, y_test_predict)
print(mse_score)
```

```python
rmse_score = root_mean_squared_error(y_test, y_test_predict)
print(rmse_score)
```

```python
R2_score = r2_score(y_test, y_test_predict)
print(R2_score)
```
